```python
def boxplot(df):
    fig, ax = plt.subplots(1, len(df.columns), figsize=(20,6))

    for i, col in enumerate(df.columns):
        df.boxplot(col, ax=ax[i])
        ax[i].set_title(col)

boxplot(df.drop(columns=['Area', 'Item']))
```

```python
def remove_outliers(df):
    for col in df.columns:
        q1 = df[col].quantile(0.25)
        q3 = df[col].quantile(.75)

        IQR = q3 - q1

        lower = q1 - IQR*1.5
        upper = q3 + IQR*1.5

        df[col] = df[col].apply(lambda x: lower if x < lower
                                          else upper if x > upper else x)
    return df
```

```python
new = remove_outliers(df.drop(columns=['Area', 'Item']))
```

```python
from sklearn.preprocessing import StandardScaler, LabelEncoder
from sklearn.metrics import r2_score, root_mean_squared_error 

from sklearn.linear_model import LinearRegression, Lasso, Ridge
from sklearn.ensemble import GradientBoostingRegressor, RandomForestRegressor
from sklearn.svm import SVR

from sklearn.model_selection import train_test_split, GridSearchCV
```

```python
no_outlier_df = new.join(df[['Area', 'Item']])
```

```python
encode = LabelEncoder()

no_outlier_df['Area'] = encode.fit_transform(no_outlier_df['Area'])
no_outlier_df['Item'] = encode.fit_transform(no_outlier_df['Item'])

no_outlier_df.info()
```

```python
scale = StandardScaler()

X = scale.fit_transform(no_outlier_df.drop('hg/ha_yield', axis=1))
y = new['hg/ha_yield']

X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=111)
```

```python
models_n_params = {
    'Linear Regression': {
        'model': LinearRegression(),
        'param_grid': {
            'fit_intercept': [True, False]
        }
    },
    'Lasso Regression': {
        'model': Lasso(random_state=111),
        'param_grid': {
            'alpha': [0.001, 0.01, 0.1, 1, 10, 100],
            'fit_intercept': [True, False]
        }
    },
    'Ridge Regression': {
        'model': Ridge(random_state=111),
        'param_grid': {
            'alpha': [0.001, 0.01, 0.1, 1, 10, 100],
            'fit_intercept': [True, False]
        }
    },
    'Gradient Boosting': {
        'model': GradientBoostingRegressor(random_state=111),
        'param_grid': {
            'n_estimators': [100, 200, 500, 1000],
            'learning_rate': [0.01, 0.05, 0.1, 0.5],
            'max_depth': [3, 5]
        }
    },
    'Random Forest': {
        'model': RandomForestRegressor(random_state=111),
        'param_grid': {
            'n_estimators': [100, 200, 500, 1000],
            'max_depth': [10, 20, None]
        }
    },
    'SVR': {
        'model': SVR(),
        'param_grid': {
            'kernel': ['linear', 'rbf'],
            'C': [0.1, 1, 10, 100],
            'gamma': [0.0001, 0.001, 0.01, 0.1, 1]
        }
    }
}
```

```python
results = []

for name, model in models_n_params.items():
    print(f'GridSearch Starting for {name}..........')

    m = model['model']
    pg = model['param_grid']

    grid = GridSearchCV(estimator=m, param_grid=pg, verbose=2, scoring=['r2', 'neg_mean_squared_error'], refit='r2')
    grid.fit(X_train, y_train)

    best_estimator = grid.best_estimator_
    prediction = best_estimator.predict(X_test)

    r2_test = r2_score(y_test, prediction)
    rmse_test = root_mean_squared_error(y_test, prediction)
    index = grid.best_index_
    best_r2_cv = grid.cv_results_['mean_test_r2'][index]
    best_mse_cv = grid.cv_results_['mean_test_neg_mean_squared_error'][index]
    best_rmse_cv = np.sqrt(best_mse_cv)

    results.append({
        'Model': name, 'Best Parameters': grid.best_params_,
        'r2_test' : r2_test, 'rmse_test' : rmse_test,
        'Best_R2_cv': best_r2_cv, 'Best_Rmse_cv' : best_rmse_cv
    })

    print(f'Done with {name}.')

```

```python
df_results = pd.DataFrame(results)
print(df_results)
```
